````
from metagpt.tools.libs.editor import Editor

editor = Editor()

complete_file = '''import { useEffect, useState, useRef, useCallback } from "react";
import { useNavigate } from "react-router-dom";
import { AppSidebar } from "@/components/AppSidebar";
import { Card } from "@/components/ui/card";
import { Button } from "@/components/ui/button";
import { Textarea } from "@/components/ui/textarea";
import {
  Select,
  SelectContent,
  SelectItem,
  SelectTrigger,
  SelectValue,
} from "@/components/ui/select";
import { client } from "@/lib/api";
import { toast } from "sonner";
import {
  Brain,
  Sparkles,
  Loader2,
  Copy,
  Check,
  History,
  ChevronDown,
  ChevronUp,
  Zap,
  MapPin,
  Users,
  TrendingUp,
  Heart,
  Shield,
  Type,
  MessageSquare,
} from "lucide-react";

/* ------------------------------------------------------------------ */
/*  Types                                                              */
/* ------------------------------------------------------------------ */

interface Product {
  id: number;
  asin: string;
  title: string;
  bullet_points: string;
  search_keywords: string;
}

interface CosmoRecord {
  id: number;
  product_id: number;
  model_name: string;
  analysis_reason: string;
  created_at: string;
}

interface DimensionKeyword {
  keyword: string;
  placement: string;
  cosmo_relation: string;
  explanation: string;
}

interface DimensionResult {
  key: string;
  title: string;
  titleEn: string;
  icon: React.ReactNode;
  color: string;
  bgColor: string;
  borderColor: string;
  keywords: DimensionKeyword[];
  summary: string;
}

type InputMode = "product" | "freetext";

/* ------------------------------------------------------------------ */
/*  6-Dimension Prompt (COSMO-Aligned with 15 Canonical Relations)     */
/* ------------------------------------------------------------------ */

const SIX_DIM_PROMPT = (productDesc: string) =>
  `You are a world-class Amazon COSMO algorithm specialist, American consumer behavior expert, and e-commerce SEO strategist. Your task is to perform a deep 6-dimensional semantic analysis of the following product, generating highly localized American English keywords that precisely align with Amazon's COSMO knowledge graph relationships.

## Product Description
${productDesc}

## COSMO Knowledge Graph Context
Amazon's COSMO system uses 15 canonical relationship types (used_for_function, capable_of, used_for_event, used_for_activity, used_when, used_where, used_with, used_for_audience, used_by, isA, has_attribute, cause_positive, cause_negative, compared_to, requires) to build a 6.3M-node knowledge graph. Your keywords must map to these relationships.

## 6 Dimensions to Analyze

### 1. FUNCTION — COSMO relations: used_for_function, capable_of, used_with
Analyze the product across these sub-dimensions:
- **Core Problem Solving**: What specific pain points does this product eliminate? (e.g., "no more tangled cords", "stops phone from dying mid-day")
- **Capabilities**: What can this product DO that consumers care about? (e.g., "fast-charges iPhone 15 to 50% in 25 min")
- **Complementary Products**: What is this commonly used WITH? (e.g., "pairs with MagSafe case", "works with any USB-C laptop")
- **Functional Differentiation**: What makes this functionally BETTER than alternatives? (e.g., "3-in-1 charging", "foldable design saves space")
- **Tech-to-Benefit Translation**: Convert technical specs into consumer-understandable benefits (e.g., "65W GaN" becomes "laptop-speed charging in your palm")
- **Multi-use Versatility**: Can it serve multiple functions? (e.g., "charges phone, tablet, AND laptop with one charger")
Generate 5-8 keywords using authentic American consumer language.

### 2. SCENARIO — COSMO relations: used_for_event, used_for_activity, used_when, used_where, used_with
Analyze the product across these sub-dimensions:
- **Activities**: Specific activities where this product is used (e.g., hiking, meal prepping, commuting, working from home, gaming)
- **Events & Occasions**: Holidays, celebrations, life events (e.g., Christmas gift, back-to-school, housewarming, baby shower)
- **American Cultural Moments**: US-specific scenarios (e.g., tailgating, road trip, dorm move-in day, Thanksgiving hosting, Super Bowl party)
- **Seasonal Context**: Time-of-year relevance (e.g., summer camping, winter storm prep, spring cleaning, fall semester)
- **Spatial Context**: WHERE specifically (e.g., tiny apartment, office cubicle, RV kitchen, gym locker, nightstand, carry-on bag)
- **Temporal Patterns**: WHEN and how often (e.g., daily commute essential, weekend warrior gear, emergency backup, overnight charging)
- **Micro-moments**: Specific trigger moments (e.g., "when your phone dies at the airport", "when you need both hands free while cooking")
Generate 5-8 keywords using authentic American consumer language.

### 3. IDENTITY — COSMO relations: used_for_audience, used_by
Analyze the product across these sub-dimensions:
- **Demographics**: Age groups (Gen Z, millennials, Gen X, boomers), gender considerations, income level signals
- **Life Stage**: College freshman, newlywed, new parent, empty nester, retiree, first-time homeowner
- **Professional Identity**: Remote worker, nurse, teacher, trucker, content creator, small business owner
- **Interest Communities**: Tech enthusiast, fitness junkie, minimalist, plant parent, pet parent, DIY-er, gamer
- **Value-Based Identity**: Eco-conscious consumer, budget-savvy shopper, premium buyer, early adopter
- **American Cultural Identity**: Military family, suburban mom, city commuter, outdoor enthusiast, college athlete
- **Gift Recipient Personas**: "gift for dad who has everything", "stocking stuffer for teens", "practical gift for college student"
Generate 5-8 keywords using authentic American consumer language.

### 4. TREND — COSMO relations: isA, has_attribute
Analyze the product across these sub-dimensions:
- **Current Aesthetic Trends**: What 2024-2025 design/style movements does this align with? (e.g., quiet luxury, clean girl aesthetic, cottagecore, dark academia, dopamine decor)
- **Social Media Trends**: TikTok Made Me Buy It, Instagram-worthy, viral product categories, influencer-driven demand
- **Material & Sustainability Trends**: Bamboo, recycled materials, plastic-free, carbon-neutral, clean ingredients, non-toxic
- **Technology Trends**: AI-powered, smart home compatible, USB-C universal, wireless/cordless, app-connected
- **Lifestyle Movement Trends**: Minimalism, self-care ritual, WFH upgrade, van life, meal prep culture, digital detox
- **Category-Specific Trends**: What is hot in THIS specific product category on Amazon right now? Rising sub-categories, emerging features
- **Color & Design Trends**: Earth tones, matte black, transparent/clear, retro revival, Scandinavian minimal
Generate 5-8 keywords using authentic American consumer language.

### 5. PSYCHOLOGY — COSMO relations: cause_positive
Analyze the product across these sub-dimensions:
- **Emotional Triggers**: Specific feelings this product creates (e.g., "peace of mind knowing your phone will not die", "that satisfying click")
- **Security & Safety**: Protection, reliability, preparedness (e.g., "never get stranded with a dead phone", "childproof design")
- **Comfort & Ease**: Convenience, simplicity, reduced friction (e.g., "one less thing to worry about", "set it and forget it")
- **Status & Social Proof**: Looking good, keeping up, being seen as smart/stylish (e.g., "the charger tech people actually use", "sleek enough for your desk setup")
- **Self-Reward & Treat Culture**: "You deserve it", upgrade mentality, small luxury (e.g., "affordable upgrade", "treat yourself to fast charging")
- **FOMO & Urgency**: Fear of missing out, anxiety relief (e.g., "everyone is switching to GaN", "stop overpaying for slow chargers")
- **Independence & Self-Reliance**: Very American values (e.g., "be your own power source", "never depend on a wall outlet again")
- **Efficiency & Optimization**: Time-saving, productivity (e.g., "charge smarter not longer", "cut your charging time in half")
Generate 5-8 keywords using authentic American consumer language.

### 6. RISK BALANCE — COSMO relations: cause_negative, compared_to, requires
Analyze the product across these sub-dimensions:
- **Durability Concerns & Reassurance**: Will it last? (e.g., "built to survive 10,000+ plug cycles", "military-grade durability")
- **Value Justification**: Is it worth the price? (e.g., "replaces 3 chargers for the price of 1", "cost per charge: $0.001")
- **Safety Certifications**: Trust signals (e.g., "FCC certified", "UL listed", "BPA-free", "FDA approved", "ETL safety tested")
- **Compatibility Worries**: Will it work with MY device? (e.g., "universal USB-C fits 99% of devices", "Apple and Android compatible")
- **Return & Warranty Confidence**: Risk-free purchase (e.g., "hassle-free 30-day returns", "lifetime warranty", "satisfaction guaranteed")
- **Competitive Comparison**: Why this over alternatives? (e.g., "half the size of Apple charger at 2x the power", "rated number 1 by Wirecutter")
- **Social Proof & Reviews**: Others experience (e.g., "50,000+ 5-star reviews", "as seen on MKBHD", "Amazons Choice")
- **Regret Prevention**: Avoiding buyers remorse (e.g., "no cheap plastic feel", "will not yellow after 3 months", "actually fits in your pocket")
Generate 5-8 keywords using authentic American consumer language.

## IMPORTANT RULES
- Generate 5-8 keywords/phrases per dimension (more coverage = better COSMO alignment)
- Use AUTHENTIC American English expressions: how real US consumers ACTUALLY search and talk on Amazon, NOT formal/textbook English
- Include colloquial phrases, slang, and natural search patterns (e.g., "charger that wont break", "best bang for your buck")
- For each keyword, specify: recommended placement (Title / Bullet Points / A+ Content / Backend Keywords)
- For each keyword, specify: the COSMO relation type it maps to (used_for_function, capable_of, used_for_event, used_for_activity, used_when, used_where, used_with, used_for_audience, used_by, isA, has_attribute, cause_positive, cause_negative, compared_to, requires)
- Include a brief explanation for why this keyword matters for Amazon SEO

## Output Format (STRICT JSON)
Return ONLY a valid JSON object with this exact structure, no markdown, no code blocks, just pure JSON:
{"function":{"keywords":[{"keyword":"...","placement":"...","cosmo_relation":"...","explanation":"..."}],"summary":"..."},"scenario":{"keywords":[{"keyword":"...","placement":"...","cosmo_relation":"...","explanation":"..."}],"summary":"..."},"identity":{"keywords":[{"keyword":"...","placement":"...","cosmo_relation":"...","explanation":"..."}],"summary":"..."},"trend":{"keywords":[{"keyword":"...","placement":"...","cosmo_relation":"...","explanation":"..."}],"summary":"..."},"psychology":{"keywords":[{"keyword":"...","placement":"...","cosmo_relation":"...","explanation":"..."}],"summary":"..."},"risk_balance":{"keywords":[{"keyword":"...","placement":"...","cosmo_relation":"...","explanation":"..."}],"summary":"..."}}`;

/* ------------------------------------------------------------------ */
/*  Dimension metadata                                                 */
/* ------------------------------------------------------------------ */

const DIMENSION_META: Record<
  string,
  {
    title: string;
    titleEn: string;
    icon: React.ReactNode;
    color: string;
    bgColor: string;
    borderColor: string;
  }
> = {
  function: {
    title: "功能",
    titleEn: "Function",
    icon: <Zap className="w-4 h-4" />,
    color: "text-blue-400",
    bgColor: "bg-blue-500/10",
    borderColor: "border-blue-500/30",
  },
  scenario: {
    title: "场景",
    titleEn: "Scenario",
    icon: <MapPin className="w-4 h-4" />,
    color: "text-emerald-400",
    bgColor: "bg-emerald-500/10",
    borderColor: "border-emerald-500/30",
  },
  identity: {
    title: "身份",
    titleEn: "Identity",
    icon: <Users className="w-4 h-4" />,
    color: "text-purple-400",
    bgColor: "bg-purple-500/10",
    borderColor: "border-purple-500/30",
  },
  trend: {
    title: "趋势",
    titleEn: "Trend",
    icon: <TrendingUp className="w-4 h-4" />,
    color: "text-amber-400",
    bgColor: "bg-amber-500/10",
    borderColor: "border-amber-500/30",
  },
  psychology: {
    title: "心理",
    titleEn: "Psychology",
    icon: <Heart className="w-4 h-4" />,
    color: "text-pink-400",
    bgColor: "bg-pink-500/10",
    borderColor: "border-pink-500/30",
  },
  risk_balance: {
    title: "风险平衡",
    titleEn: "Risk Balance",
    icon: <Shield className="w-4 h-4" />,
    color: "text-red-400",
    bgColor: "bg-red-500/10",
    borderColor: "border-red-500/30",
  },
};

const PLACEMENT_COLORS: Record<string, string> = {
  Title: "bg-blue-600/80 text-white",
  "Bullet Points": "bg-emerald-600/80 text-white",
  "A+ Content": "bg-purple-600/80 text-white",
  "Backend Keywords": "bg-gray-600/80 text-white",
};

const COSMO_RELATION_COLORS: Record<string, string> = {
  used_for_function: "bg-blue-600/30 text-blue-300",
  capable_of: "bg-cyan-600/30 text-cyan-300",
  used_for_event: "bg-emerald-600/30 text-emerald-300",
  used_for_activity: "bg-green-600/30 text-green-300",
  used_when: "bg-teal-600/30 text-teal-300",
  used_where: "bg-lime-600/30 text-lime-300",
  used_with: "bg-yellow-600/30 text-yellow-300",
  used_for_audience: "bg-purple-600/30 text-purple-300",
  used_by: "bg-violet-600/30 text-violet-300",
  isA: "bg-amber-600/30 text-amber-300",
  has_attribute: "bg-orange-600/30 text-orange-300",
  cause_positive: "bg-pink-600/30 text-pink-300",
  cause_negative: "bg-red-600/30 text-red-300",
  compared_to: "bg-rose-600/30 text-rose-300",
  requires: "bg-slate-600/30 text-slate-300",
};

/* ------------------------------------------------------------------ */
/*  CopyButton                                                         */
/* ------------------------------------------------------------------ */

function CopyButton({
  text,
  size = "sm",
}: {
  text: string;
  size?: "sm" | "lg";
}) {
  const [copied, setCopied] = useState(false);
  const handleCopy = () => {
    navigator.clipboard.writeText(text).then(() => {
      setCopied(true);
      toast.success("已复制到剪贴板");
      setTimeout(() => setCopied(false), 2000);
    });
  };
  return (
    <Button
      variant="ghost"
      size="sm"
      onClick={handleCopy}
      className={`text-gray-400 hover:text-white ${size === "lg" ? "h-8 px-3" : "h-7 px-2"}`}
    >
      {copied ? (
        <Check className="w-3.5 h-3.5 text-emerald-400" />
      ) : (
        <Copy className="w-3.5 h-3.5" />
      )}
      {size === "lg" && (
        <span className="ml-1 text-xs">{copied ? "已复制" : "复制"}</span>
      )}
    </Button>
  );
}

/* ------------------------------------------------------------------ */
/*  DimensionCard                                                      */
/* ------------------------------------------------------------------ */

function DimensionCard({ dim }: { dim: DimensionResult }) {
  const allText = dim.keywords
    .map((k) => `${k.keyword} [${k.placement}] (${k.cosmo_relation}) - ${k.explanation}`)
    .join("\\n");

  return (
    <Card
      className={`${dim.bgColor} ${dim.borderColor} border p-4 sm:p-5 transition-all duration-300 hover:scale-[1.01] hover:shadow-lg`}
    >
      {/* Header */}
      <div className="flex items-center justify-between mb-3">
        <div className="flex items-center gap-2">
          <div
            className={`w-8 h-8 rounded-lg ${dim.bgColor} flex items-center justify-center ${dim.color}`}
          >
            {dim.icon}
          </div>
          <div>
            <h3 className={`font-bold text-sm ${dim.color}`}>{dim.title}</h3>
            <p className="text-[10px] text-gray-500">{dim.titleEn}</p>
          </div>
        </div>
        <CopyButton text={allText} size="lg" />
      </div>

      {/* Summary */}
      {dim.summary && (
        <p className="text-xs text-gray-400 mb-3 leading-relaxed border-b border-white/5 pb-3">
          {dim.summary}
        </p>
      )}

      {/* Keywords */}
      <div className="space-y-2.5">
        {dim.keywords.map((kw, idx) => (
          <div
            key={idx}
            className="group rounded-lg bg-white/[0.03] hover:bg-white/[0.06] p-2.5 transition-colors"
          >
            <div className="flex items-start justify-between gap-2">
              <div className="flex-1 min-w-0">
                <div className="flex items-center gap-2 flex-wrap">
                  <span className="text-white font-medium text-sm">
                    {kw.keyword}
                  </span>
                  <span
                    className={`text-[10px] px-1.5 py-0.5 rounded-full font-medium ${PLACEMENT_COLORS[kw.placement] || "bg-gray-600/80 text-white"}`}
                  >
                    {kw.placement}
                  </span>
                  {kw.cosmo_relation && (
                    <span
                      className={`text-[9px] px-1.5 py-0.5 rounded-full font-mono ${COSMO_RELATION_COLORS[kw.cosmo_relation] || "bg-indigo-600/30 text-indigo-300"}`}
                    >
                      {kw.cosmo_relation}
                    </span>
                  )}
                </div>
                <p className="text-xs text-gray-500 mt-1 leading-relaxed">
                  {kw.explanation}
                </p>
              </div>
              <CopyButton text={kw.keyword} />
            </div>
          </div>
        ))}
      </div>
    </Card>
  );
}

/* ------------------------------------------------------------------ */
/*  Main Component                                                     */
/* ------------------------------------------------------------------ */

export default function CosmoTranslator() {
  const navigate = useNavigate();
  const [products, setProducts] = useState<Product[]>([]);
  const [selectedProductId, setSelectedProductId] = useState<string>("");
  const [freeText, setFreeText] = useState("");
  const [inputMode, setInputMode] = useState<InputMode>("freetext");
  const [loading, setLoading] = useState(false);
  const [dimensions, setDimensions] = useState<DimensionResult[]>([]);
  const [rawResponse, setRawResponse] = useState("");
  const [historyRecords, setHistoryRecords] = useState<CosmoRecord[]>([]);
  const [showHistory, setShowHistory] = useState(false);
  const rawRef = useRef("");

  useEffect(() => {
    checkAuth();
  }, []);

  const checkAuth = async () => {
    try {
      const u = await client.auth.me();
      if (!u?.data) {
        navigate("/");
        return;
      }
    } catch {
      navigate("/");
      return;
    }
    // Load products and history separately so errors don't kick user out
    try {
      const res = await client.entities.products.query({
        sort: "-created_at",
        limit: 50,
      });
      setProducts(res?.data?.items || []);
    } catch (e) {
      console.error("Failed to load products:", e);
    }
    loadHistory();
  };

  const loadHistory = async () => {
    try {
      const res = await client.entities.cosmo_results.query({
        sort: "-created_at",
        limit: 20,
      });
      setHistoryRecords(res?.data?.items || []);
    } catch (e) {
      console.error(e);
    }
  };

  const selectedProduct = products.find(
    (p) => String(p.id) === selectedProductId
  );

  /* ---- Build product description for the prompt ---- */
  const getProductDescription = useCallback((): string => {
    if (inputMode === "freetext") {
      return freeText.trim();
    }
    if (!selectedProduct) return "";
    const parts = [`Product: ${selectedProduct.title}`];
    if (selectedProduct.bullet_points)
      parts.push(`Bullet Points: ${selectedProduct.bullet_points}`);
    if (selectedProduct.search_keywords)
      parts.push(`Search Keywords: ${selectedProduct.search_keywords}`);
    return parts.join("\\n");
  }, [inputMode, freeText, selectedProduct]);

  /* ---- Parse JSON from AI response ---- */
  const parseResponse = useCallback((text: string): DimensionResult[] => {
    try {
      // Try to extract JSON from the response
      let jsonStr = text.trim();

      // Remove markdown code block if present
      const jsonMatch = jsonStr.match(/```(?:json)?\\s*([\\s\\S]*?)```/);
      if (jsonMatch) {
        jsonStr = jsonMatch[1].trim();
      }

      // Try to find JSON object boundaries
      const startIdx = jsonStr.indexOf("{");
      const endIdx = jsonStr.lastIndexOf("}");
      if (startIdx !== -1 && endIdx !== -1 && endIdx > startIdx) {
        jsonStr = jsonStr.substring(startIdx, endIdx + 1);
      }

      const parsed = JSON.parse(jsonStr);
      const results: DimensionResult[] = [];

      const dimKeys = [
        "function",
        "scenario",
        "identity",
        "trend",
        "psychology",
        "risk_balance",
      ];

      for (const key of dimKeys) {
        const meta = DIMENSION_META[key];
        const data = parsed[key];
        if (meta && data) {
          results.push({
            key,
            title: meta.title,
            titleEn: meta.titleEn,
            icon: meta.icon,
            color: meta.color,
            bgColor: meta.bgColor,
            borderColor: meta.borderColor,
            keywords: (data.keywords || []).map((k: any) => ({
              keyword: k.keyword || "",
              placement: k.placement || "Backend Keywords",
              cosmo_relation: k.cosmo_relation || "",
              explanation: k.explanation || "",
            })),
            summary: data.summary || "",
          });
        }
      }

      return results;
    } catch {
      return [];
    }
  }, []);

  /* ---- Run 6-dimension analysis ---- */
  const runAnalysis = async () => {
    const desc = getProductDescription();
    if (!desc) {
      toast.error(
        inputMode === "freetext"
          ? "请输入产品描述"
          : "请先选择一个产品"
      );
      return;
    }

    setLoading(true);
    setDimensions([]);
    setRawResponse("");
    rawRef.current = "";

    const prompt = SIX_DIM_PROMPT(desc);

    try {
      await client.ai.gentxt({
        messages: [
          {
            role: "system",
            content:
              "You are an Amazon COSMO algorithm expert with deep knowledge of COSMO\\'s 15 canonical relationship types (used_for_function, capable_of, used_for_event, used_for_activity, used_when, used_where, used_with, used_for_audience, used_by, isA, has_attribute, cause_positive, cause_negative, compared_to, requires). You specialize in American consumer language localization and Amazon SEO. Always respond with valid JSON only, no markdown formatting.",
          },
          { role: "user", content: prompt },
        ],
        model: "claude-4-5-sonnet",
        stream: true,
        onChunk: (chunk: any) => {
          rawRef.current += chunk.content || "";
          setRawResponse(rawRef.current);
        },
        onComplete: () => {
          setLoading(false);
          const parsed = parseResponse(rawRef.current);
          if (parsed.length > 0) {
            setDimensions(parsed);
            toast.success("6维语义分析完成！");
          } else {
            toast.error("解析结果失败，请重试");
          }

          // Save to history — use product_id 0 for freetext mode
          const productId = selectedProduct ? selectedProduct.id : 0;
          client.entities.cosmo_results
            .create({
              data: {
                product_id: productId,
                model_name: "claude-4-5-sonnet",
                optimized_title: desc.substring(0, 500),
                optimized_bullets: "",
                optimized_keywords: "",
                analysis_reason: rawRef.current,
              },
            })
            .then(() => loadHistory())
            .catch((e: any) => console.error("Failed to save history:", e));
        },
        onError: (err: any) => {
          setLoading(false);
          toast.error("分析失败: " + (err?.message || "未知错误"));
        },
      });
    } catch (err: any) {
      setLoading(false);
      toast.error("分析请求失败: " + (err?.message || "未知错误"));
    }
  };

  /* ---- Copy all dimensions ---- */
  const copyAllDimensions = () => {
    if (dimensions.length === 0) return;
    const text = dimensions
      .map((dim) => {
        const header = `【${dim.title} / ${dim.titleEn}】`;
        const kwLines = dim.keywords
          .map(
            (k) => `  • ${k.keyword} [${k.placement}] (${k.cosmo_relation}) — ${k.explanation}`
          )
          .join("\\n");
        return `${header}\\n${dim.summary}\\n${kwLines}`;
      })
      .join("\\n\\n");
    navigator.clipboard.writeText(text).then(() => {
      toast.success("已复制全部6维分析结果");
    });
  };

  return (
    <div className="flex h-screen bg-[#060A14] text-white">
      <AppSidebar />
      <main className="flex-1 overflow-y-auto">
        <div className="p-4 sm:p-6 max-w-7xl mx-auto pt-14 md:pt-6">
          {/* Header */}
          <div className="flex flex-col sm:flex-row sm:items-center justify-between mb-6 sm:mb-8 gap-3">
            <div>
              <h1 className="text-xl sm:text-2xl font-bold flex items-center gap-2">
                <Brain className="w-5 h-5 sm:w-6 sm:h-6 text-indigo-400" />
                Cosmo 6维语义引擎
              </h1>
              <p className="text-gray-400 mt-1 text-sm">
                基于 COSMO 15种规范关系的 6 维度深度语义分析 · Powered by Claude
              </p>
            </div>
            <div className="flex items-center gap-2">
              {dimensions.length > 0 && (
                <Button
                  variant="outline"
                  size="sm"
                  onClick={copyAllDimensions}
                  className="border-white/10 text-gray-300 hover:text-white hover:bg-white/5 bg-transparent"
                >
                  <Copy className="w-4 h-4 mr-1.5" /> 复制全部
                </Button>
              )}
              <Button
                variant="outline"
                size="sm"
                onClick={() => setShowHistory(!showHistory)}
                className="border-white/10 text-gray-300 hover:text-white hover:bg-white/5 bg-transparent"
              >
                <History className="w-4 h-4 mr-1.5" /> 历史{" "}
                {showHistory ? (
                  <ChevronUp className="w-3 h-3 ml-1" />
                ) : (
                  <ChevronDown className="w-3 h-3 ml-1" />
                )}
              </Button>
            </div>
          </div>

          {/* History Panel */}
          {showHistory && (
            <Card className="bg-[#0D1321] border-white/[0.06] p-4 mb-6 max-h-48 overflow-y-auto">
              <h3 className="text-xs font-semibold text-gray-500 uppercase mb-3">
                最近分析记录
              </h3>
              {historyRecords.length === 0 ? (
                <p className="text-xs text-gray-600 text-center py-2">暂无历史记录</p>
              ) : (
                <div className="space-y-2">
                  {historyRecords.slice(0, 10).map((r) => {
                    const prod = products.find((p) => p.id === r.product_id);
                    const isFreetext = !r.product_id || r.product_id === 0;
                    return (
                      <div
                        key={r.id}
                        className="flex items-center justify-between text-xs p-2 rounded-lg bg-white/[0.02] hover:bg-white/[0.04] transition-colors"
                      >
                        <div className="flex items-center gap-2 min-w-0">
                          <span className={`font-mono flex-shrink-0 ${isFreetext ? "text-emerald-400" : "text-indigo-400"}`}>
                            {isFreetext ? "自由输入" : (prod?.asin || `#${r.product_id}`)}
                          </span>
                          <span className="text-gray-500 truncate">
                            {r.model_name}
                          </span>
                        </div>
                        <span className="text-gray-600 flex-shrink-0">
                          {r.created_at
                            ? new Date(r.created_at).toLocaleDateString()
                            : ""}
                        </span>
                      </div>
                    );
                  })}
                </div>
              )}
            </Card>
          )}

          {/* Input Section */}
          <Card className="bg-[#0D1321] border-white/[0.06] p-4 sm:p-5 mb-6">
            {/* Mode Tabs */}
            <div className="flex gap-2 mb-4">
              <button
                onClick={() => setInputMode("freetext")}
                className={`flex items-center gap-1.5 px-3 py-1.5 rounded-lg text-sm font-medium transition-all ${
                  inputMode === "freetext"
                    ? "bg-indigo-600/20 text-indigo-400 border border-indigo-500/30"
                    : "text-gray-500 hover:text-gray-300 border border-transparent"
                }`}
              >
                <Type className="w-3.5 h-3.5" /> 自由输入
              </button>
              <button
                onClick={() => setInputMode("product")}
                className={`flex items-center gap-1.5 px-3 py-1.5 rounded-lg text-sm font-medium transition-all ${
                  inputMode === "product"
                    ? "bg-indigo-600/20 text-indigo-400 border border-indigo-500/30"
                    : "text-gray-500 hover:text-gray-300 border border-transparent"
                }`}
              >
                <MessageSquare className="w-3.5 h-3.5" /> 选择产品
              </button>
            </div>

            {/* Input Area */}
            {inputMode === "freetext" ? (
              <div className="space-y-3">
                <Textarea
                  value={freeText}
                  onChange={(e) => setFreeText(e.target.value)}
                  placeholder="输入产品描述，例如：手机充电器、USB-C快充、65W氮化镓充电头..."
                  className="bg-white/5 border-white/10 text-white placeholder:text-gray-600 min-h-[80px] resize-none focus:border-indigo-500/50"
                />
                <div className="flex flex-wrap gap-2">
                  {[
                    "手机充电器 USB-C 快充",
                    "瑜伽垫 防滑 加厚",
                    "儿童水杯 保温 316不锈钢",
                    "车载手机支架 磁吸",
                  ].map((example) => (
                    <button
                      key={example}
                      onClick={() => setFreeText(example)}
                      className="text-xs px-2.5 py-1 rounded-full bg-white/5 text-gray-500 hover:text-gray-300 hover:bg-white/10 transition-colors border border-white/5"
                    >
                      {example}
                    </button>
                  ))}
                </div>
              </div>
            ) : (
              <Select
                value={selectedProductId}
                onValueChange={setSelectedProductId}
              >
                <SelectTrigger className="bg-white/5 border-white/10 text-white">
                  <SelectValue placeholder="选择一个ASIN产品" />
                </SelectTrigger>
                <SelectContent className="bg-[#1F2937] border-white/10">
                  {products.map((p) => (
                    <SelectItem
                      key={p.id}
                      value={String(p.id)}
                      className="text-white hover:bg-white/10"
                    >
                      {p.asin} - {p.title.substring(0, 50)}
                    </SelectItem>
                  ))}
                </SelectContent>
              </Select>
            )}

            {/* Action Button */}
            <div className="mt-4">
              <Button
                onClick={runAnalysis}
                disabled={loading}
                className="w-full sm:w-auto bg-gradient-to-r from-indigo-600 to-purple-600 hover:from-indigo-500 hover:to-purple-500 text-white shadow-lg shadow-indigo-500/20 h-10"
              >
                {loading ? (
                  <>
                    <Loader2 className="w-4 h-4 mr-1.5 animate-spin" />{" "}
                    6维分析中...
                  </>
                ) : (
                  <>
                    <Sparkles className="w-4 h-4 mr-1.5" /> 启动 6维语义分析
                  </>
                )}
              </Button>
            </div>
          </Card>

          {/* 6-Dimension Legend */}
          {(dimensions.length > 0 || loading) && (
            <div className="flex flex-wrap gap-2 mb-4">
              {Object.entries(DIMENSION_META).map(([key, meta]) => (
                <div
                  key={key}
                  className={`flex items-center gap-1.5 text-xs ${meta.color} opacity-80`}
                >
                  {meta.icon}
                  <span>
                    {meta.title}/{meta.titleEn}
                  </span>
                </div>
              ))}
            </div>
          )}

          {/* Placement Legend */}
          {dimensions.length > 0 && (
            <div className="flex flex-wrap gap-2 mb-6">
              <span className="text-xs text-gray-600 mr-1">推荐位置：</span>
              {Object.entries(PLACEMENT_COLORS).map(([label, cls]) => (
                <span
                  key={label}
                  className={`text-[10px] px-2 py-0.5 rounded-full ${cls}`}
                >
                  {label}
                </span>
              ))}
              <span className="text-xs text-gray-600 ml-3 mr-1">COSMO关系：</span>
              <span className="text-[9px] px-1.5 py-0.5 rounded-full bg-indigo-600/30 text-indigo-300 font-mono">
                relation_type
              </span>
            </div>
          )}

          {/* Loading State - Streaming */}
          {loading && rawResponse && (
            <Card className="bg-[#0D1321] border-white/[0.06] p-4 sm:p-5 mb-6">
              <div className="flex items-center gap-2 mb-3">
                <Loader2 className="w-4 h-4 animate-spin text-indigo-400" />
                <span className="text-sm text-indigo-400 font-medium">
                  AI 正在进行 6维深度语义分析...
                </span>
              </div>
              <pre className="whitespace-pre-wrap text-xs text-gray-500 leading-relaxed font-mono max-h-40 overflow-y-auto">
                {rawResponse.length > 500
                  ? "..." + rawResponse.slice(-500)
                  : rawResponse}
              </pre>
            </Card>
          )}

          {/* 6-Dimension Results Grid */}
          {dimensions.length > 0 && (
            <div className="grid grid-cols-1 md:grid-cols-2 xl:grid-cols-3 gap-4 sm:gap-5">
              {dimensions.map((dim) => (
                <DimensionCard key={dim.key} dim={dim} />
              ))}
            </div>
          )}

          {/* Empty State */}
          {!loading && dimensions.length === 0 && !rawResponse && (
            <Card className="bg-[#0D1321] border-white/[0.06] p-8 sm:p-12 text-center">
              <div className="relative mx-auto w-20 h-20 mb-6">
                <div className="absolute inset-0 rounded-full bg-gradient-to-br from-indigo-500/20 to-purple-500/20 animate-pulse" />
                <div className="absolute inset-2 rounded-full bg-[#0D1321] flex items-center justify-center">
                  <Brain className="w-8 h-8 text-gray-600" />
                </div>
              </div>
              <h3 className="text-gray-400 font-medium mb-2">
                6维 COSMO 语义引擎
              </h3>
              <p className="text-gray-600 text-sm max-w-md mx-auto leading-relaxed">
                输入产品描述，AI 将基于 Amazon COSMO 的 15 种规范关系类型，
                从功能、场景、身份、趋势、心理、风险平衡 6 个维度深度分析，
                每个关键词标注 COSMO 关系类型和推荐位置，精准对齐算法语义
              </p>

              {/* 6-Dimension Preview */}
              <div className="grid grid-cols-2 sm:grid-cols-3 gap-3 mt-8 max-w-lg mx-auto">
                {Object.entries(DIMENSION_META).map(([key, meta]) => (
                  <div
                    key={key}
                    className={`${meta.bgColor} ${meta.borderColor} border rounded-lg p-3 text-center`}
                  >
                    <div
                      className={`${meta.color} flex items-center justify-center mb-1`}
                    >
                      {meta.icon}
                    </div>
                    <p className={`text-xs font-medium ${meta.color}`}>
                      {meta.title}
                    </p>
                    <p className="text-[10px] text-gray-600">
                      {meta.titleEn}
                    </p>
                  </div>
                ))}
              </div>

              {/* COSMO Relations Preview */}
              <div className="mt-6 max-w-2xl mx-auto">
                <p className="text-[10px] text-gray-600 mb-2">覆盖 COSMO 15种规范关系类型</p>
                <div className="flex flex-wrap justify-center gap-1.5">
                  {Object.entries(COSMO_RELATION_COLORS).map(([rel, cls]) => (
                    <span key={rel} className={`text-[9px] px-1.5 py-0.5 rounded-full font-mono ${cls}`}>
                      {rel}
                    </span>
                  ))}
                </div>
              </div>
            </Card>
          )}
        </div>
      </main>
    </div>
  );
}'''

await editor.write("/workspace/app/frontend/src/pages/CosmoTranslator.tsx", complete_file)
print("✅ CosmoTranslator.tsx completely rewritten with COSMO-aligned 6-dimension engine!")

# Verify the file
content = await editor.read("/workspace/app/frontend/src/pages/CosmoTranslator.tsx")
text = str(content)
lines = text.split('\n')
print(f"📊 Total lines: {len(lines)}")

# Quick verification checks
checks = {
    "COSMO Knowledge Graph Context": "COSMO Knowledge Graph Context" in text,
    "15 canonical relationship types": "15 canonical relationship types" in text,
    "cosmo_relation field": "cosmo_relation: string;" in text,
    "Core Problem Solving": "Core Problem Solving" in text,
    "American Cultural Moments": "American Cultural Moments" in text,
    "Gift Recipient Personas": "Gift Recipient Personas" in text,
    "Social Media Trends": "Social Media Trends" in text,
    "FOMO & Urgency": "FOMO" in text,
    "Regret Prevention": "Regret Prevention" in text,
    "COSMO_RELATION_COLORS": "COSMO_RELATION_COLORS" in text,
    "Generate 5-8 keywords": "Generate 5-8 keywords" in text,
    "claude-4-5-sonnet": "claude-4-5-sonnet" in text,
}

all_pass = True
for check, result in checks.items():
    status = "✅" if result else "❌"
    if not result:
        all_pass = False
    print(f"  {status} {check}")

if all_pass:
    print("\n🎉 All checks passed! File is complete and correct.")
else:
    print("\n⚠️ Some checks failed, review needed.")
````
